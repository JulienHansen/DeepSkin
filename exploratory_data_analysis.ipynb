```python
import os 
import matplotlib.pyplot as plt
import torch
import pandas as pd 
import seaborn as sns
import warnings

from torch.utils.data import  Dataset, DataLoader
from torchvision import transforms
from PIL import Image
from matplotlib.colors import ListedColormap
```

## Data Import


```python
data = pd.read_csv('archive/HAM10000_metadata.csv')

images_path_1 = 'archive/HAM10000_images_part_1/'
images_path_2 = 'archive/HAM10000_images_part_2/'

# Create a column 'image_path' that points to the actual image
data['image_path'] = data['image_id'].apply(
    lambda x: os.path.join(images_path_1, x + '.jpg')
    if os.path.exists(os.path.join(images_path_1, x + '.jpg'))
    else os.path.join(images_path_2, x + '.jpg')
)
print(data.head())
```

```python
class HAM10000ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        """
        Args:
            dataframe (pd.DataFrame): DataFrame containing image paths and labels.
            transform (callable, optional): Transform to be applied to the images.
        """
        self.data = dataframe.reset_index(drop=True)
        # Use the provided transform or default to converting the image to a tensor.
        self.transform = transform if transform is not None else transforms.ToTensor()
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        image_path = self.data.loc[idx, 'image_path']
        # Open the image and ensure it is in RGB format.
        image = Image.open(image_path).convert('RGB')
        if self.transform:
            image = self.transform(image)
        label = self.data.loc[idx, 'dx']
        return image, label
```

## Mean and standard deviation in the dataset

```python
# Create an instance of the dataset
dataset = HAM10000ImageDataset(data, transform=transforms.ToTensor())

def get_mean_std(dataset, batch_size):
    dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=False)
    mean = torch.zeros(3)
    std = torch.zeros(3)
    nb_samples = 0
    for images, _ in dataloader:
        batch_samples = images.size(0)
        # Reshape images to (batch, channels, height*width)
        images = images.view(batch_samples, images.size(1), -1)
        mean += images.mean(2).sum(0)
        std += images.std(2).sum(0)
        nb_samples += batch_samples
    mean /= nb_samples
    std /= nb_samples
    return mean, std

# Compute and print the mean and standard deviation
mean, std = get_mean_std(dataset, batch_size=256)
print('Mean:', mean)
print('Std:', std)
```

# Correlation matrices

```python
# Convert categorical variables into numerical values
df_encoded = data.copy()
df_encoded = df_encoded.drop(columns=['lesion_id', 'image_id', 'image_path'])

# Store original category labels
dx_labels = df_encoded['dx'].unique()
dx_type_labels = df_encoded['dx_type'].unique()
sex_labels = df_encoded['sex'].unique()
localization_labels = df_encoded['localization'].unique()

df_encoded["sex"] = df_encoded["sex"].map({"male": 0, "female": 1})
df_encoded["dx_type"] = df_encoded["dx_type"].astype("category").cat.codes
df_encoded["localization"] = df_encoded["localization"].astype("category").cat.codes
df_encoded["dx"] = df_encoded["dx"].astype("category").cat.codes

# Compute correlation matrix
corr_matrix = df_encoded.corr()

# Plot correlation matrix
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Dataset")
plt.show()
```

```python
# sns.pairplot(df_encoded, diag_kind='kde', markers="o", plot_kws={'alpha': 0.6})
# plt.show()

# Set style for readability
sns.set_theme(style="whitegrid", font_scale=1.5)

# Create a scatter matrix with better readability
pairplot = sns.pairplot(
    df_encoded, 
    diag_kind='kde',         # Kernel Density Estimation for diagonal plots
    hue='dx',                # Color by lesion type for better grouping
    palette='viridis',       # Use a color palette for distinction
    markers='o',             # Use circles as markers
    plot_kws={'alpha': 0.6, 's': 40},  # Transparency and smaller markers
    height=3,              # Increase figure size for clarity
    aspect=1.3   # 🔥 Stretch subplots horizontally
)

# Function to replace axis labels with original category names
def replace_axis_labels(g, column, labels):
    for ax in g.axes.flat:
        if ax.get_xlabel() == column:
            ax.set_xticks(range(len(labels)))
            ax.set_xticklabels(labels, rotation=60, ha="right", fontsize=12)
        if ax.get_ylabel() == column:
            ax.set_yticks(range(len(labels)))
            ax.set_yticklabels(labels, rotation=45, ha="right", fontsize=12)

# Apply the label replacement
replace_axis_labels(pairplot, 'dx', dx_labels)
replace_axis_labels(pairplot, 'dx_type', dx_type_labels)
replace_axis_labels(pairplot, 'sex', sex_labels)
replace_axis_labels(pairplot, 'localization', localization_labels)

print(localization_labels)

# Modify legend to include dx labels
legend = pairplot._legend
legend.set_title("dx (Diagnosis)")
for t, label in zip(legend.texts, dx_labels):
    t.set_text(label)

plt.savefig("scatter_matrix.png", dpi=300, bbox_inches='tight')

# # Adjust layout for better spacing
# plt.tight_layout()

# Show the plot
plt.show()
```
